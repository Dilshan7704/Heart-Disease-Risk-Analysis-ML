```python
import pandas as pd
import joblib
import warnings

warnings.filterwarnings('ignore')
```

```python
# STEP 1 — LOAD PATIENT DATA

patient_data = {
    'Age': 58,
    'Sex': 1,
    'Chest pain type': 1,
    'BP': 125,
    'Cholesterol': 325,
    'FBS over 120': 0,
    'EKG results': 2,
    'Max HR': 171,
    'Exercise angina': 0,
    'ST depression': 0.000,
    'Slope of ST': 1,
    'Number of vessels fluro': 0,
    'Thallium': 3
}
```

```python
# Convert to DataFrame
input_df = pd.DataFrame([patient_data])
model_path = '../Models/with_out_pca_models/CatBoost_model.pkl'
model = joblib.load(model_path)
print("CatBoost Model Loaded Successfully!")
```

```python
print("\n" + "="*60)
print("❤️ HEART DISEASE PREDICTION SYSTEM")
print("="*60)

try:

    # Get probability score
    if hasattr(model, "predict_proba"):
        probability = model.predict_proba(input_df)[0][1]
    else:
        probability = model.decision_function(input_df)[0]

    # Decision Threshold
    threshold = 0.5

    # Final Prediction
    prediction = 1 if probability > threshold else 0


    if prediction == 1:
        result = "🔴🫀⚠️ HIGH RISK OF HEART DISEASE"
        status = "Presence Detected"
    else:
        result = "🟢🫀🛡️ LOW RISK OF HEART DISEASE"
        status = "No Heart Disease Detected"

    print("\n📊 PREDICTION RESULTS\n")

    print(f"🫀 Prediction Status      : {status}")
    print(f"📈 Risk Probability       : {probability * 100:.2f}%")
    print(f"🎯 Decision Threshold     : {threshold}")

    print(f"{result}")

    if prediction == 1:

        print("\n⚠️ Medical Recommendation:")
        print("Patient may require further cardiac evaluation,")
        print("clinical diagnosis, and medical consultation.")

    else:

        print("\n🩺🫀 Medical Recommendation:")
        print("Patient currently shows low heart disease risk.")
        print("Maintain healthy lifestyle and regular checkups.")

except Exception as e:

    print("\n Prediction Failed!")
    print("Error:", e)
```
